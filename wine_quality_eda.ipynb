# Exploratory Data Analysis: Wine Quality Dataset

This project analyzes the Wine Quality dataset using Python.

The goal is to explore distributions, descriptive statistics,
outliers, and relationships between different variables.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the Wine Quality dataset
wine = pd.read_csv("data/winequality-red.csv", sep=";")

# Display the first five rows
wine.head()
```

```python
print("Rows and columns:", wine.shape)
print("\nColumn names:")
print(wine.columns.tolist())
```

```python
# Check data types
wine.info()

# Check missing values
print("\nMissing values:")
print(wine.isnull().sum())

# Check duplicate rows
print("\nDuplicate rows:", wine.duplicated().sum())
```

```python
# Display summary statistics
wine.describe().T
```

```python
# Calculate outliers using the IQR method
Q1 = wine.quantile(0.25)
Q3 = wine.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = ((wine < lower_bound) | (wine > upper_bound)).sum()

# Summary table
summary = wine.describe().T[["min", "max", "mean", "50%"]]
summary = summary.rename(columns={"50%": "median"})
summary["outlier_count"] = outliers

display(summary)
```

## Descriptive Statistics and Outlier Analysis

The dataset was examined for missing values, duplicate observations, and the distribution of numerical attributes.

Descriptive statistics were calculated to identify the minimum, maximum, mean, and median of each variable.

Potential outliers were identified using the Interquartile Range (IQR) method. Values below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR were classified as potential outliers.

Outliers were retained because they may represent meaningful variations in the chemical properties of wine.

## Univariate Analysis: Distributions of Wine Attributes

Histograms are used to examine the distribution of each numerical variable, including its shape, spread, and potential outliers.

```python
# Histograms for all numerical variables
wine.hist(figsize=(16, 14), bins=25, edgecolor="black")

plt.suptitle("Distribution of Wine Quality Attributes", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# Combined box plot of all numerical variables
plt.figure(figsize=(16, 7))

sns.boxplot(data=wine)

plt.title("Box Plots of All Wine Attributes")
plt.xlabel("Wine Attributes")
plt.ylabel("Values")
plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```

```python
# Standardize numerical variables for easier comparison
wine_scaled = (wine - wine.mean()) / wine.std()

plt.figure(figsize=(16, 7))

sns.boxplot(data=wine_scaled)

plt.title("Standardized Box Plots of Wine Attributes")
plt.xlabel("Wine Attributes")
plt.ylabel("Standardized Values (Z-scores)")
plt.xticks(rotation=90)

plt.tight_layout()
plt.show()
```

### Distribution Analysis

The histograms show how the chemical properties and quality scores of red wine are distributed.

Some attributes appear approximately symmetric, while others are skewed, with most observations concentrated at lower values and a smaller number of unusually high measurements.

The combined box plot highlights differences in the ranges and variability of the attributes. Because the variables use different measurement scales, a standardized box plot was also created to make comparisons easier.

Several attributes contain potential outliers, as indicated by observations beyond the box plot whiskers. These observations are retained for further analysis.

## Bivariate Analysis: Numerical Relationships

Scatter plots help identify relationships between two numerical variables.

A scatter plot matrix allows us to compare multiple numerical attributes simultaneously, while a correlation heatmap measures the strength and direction of their linear relationships.

```python
# Select numerical variables
numerical_columns = wine.select_dtypes(include="number")

# Create a scatter plot matrix
sns.pairplot(numerical_columns, diag_kind="hist",
             plot_kws={"alpha": 0.3, "s": 10})

plt.show()
```

```python
# Calculate Pearson correlation coefficients
correlation_matrix = wine.corr(numeric_only=True)

plt.figure(figsize=(12, 10))

sns.heatmap(correlation_matrix,
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("Correlation Heatmap of Wine Attributes")
plt.tight_layout()
plt.show()
```

```python
# Correlation of each attribute with wine quality
quality_correlations = correlation_matrix["quality"].sort_values(
    ascending=False
)

print(quality_correlations)
```

### Numerical Relationship Analysis

The scatter plot matrix visualizes pairwise relationships between the numerical attributes of the wine dataset.

The Pearson correlation heatmap provides a numerical measure of these relationships, with values ranging from -1 to 1.

Positive correlations indicate that two variables tend to increase together, while negative correlations indicate that one variable tends to decrease as the other increases.

The correlation analysis also helps identify which chemical properties are most strongly associated with wine quality. However, correlation does not necessarily imply causation.

## Categorical Data Analysis

To examine categorical relationships, two new variables are created:

- Wine Quality Category: Low, Medium, or High
- Alcohol Category: Low, Medium, or High

These categories allow us to compare groups and investigate relationships between categorical and numerical variables.

```python
# Categorize wine quality
wine["quality_category"] = pd.cut(
    wine["quality"],
    bins=[0, 4, 6, 10],
    labels=["Low", "Medium", "High"]
)

# Categorize alcohol content
wine["alcohol_category"] = pd.cut(
    wine["alcohol"],
    bins=[0, 10, 12, float("inf")],
    labels=["Low", "Medium", "High"]
)

# Display category counts
print(wine["quality_category"].value_counts())
print(wine["alcohol_category"].value_counts())
```

```python
# Cross-tabulate quality and alcohol categories
category_table = pd.crosstab(
    wine["quality_category"],
    wine["alcohol_category"]
)

display(category_table)

# Grouped bar chart
category_table.plot(kind="bar", figsize=(9, 6))

plt.title("Wine Quality Category vs Alcohol Category")
plt.xlabel("Wine Quality Category")
plt.ylabel("Number of Wines")
plt.xticks(rotation=0)
plt.legend(title="Alcohol Category")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9, 6))

sns.boxplot(
    data=wine,
    x="quality_category",
    y="alcohol"
)

plt.title("Alcohol Content by Wine Quality Category")
plt.xlabel("Wine Quality Category")
plt.ylabel("Alcohol Content (%)")

plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=wine,
    x="alcohol",
    y="volatile acidity",
    hue="quality_category",
    alpha=0.6
)

plt.title("Alcohol vs Volatile Acidity by Wine Quality")
plt.xlabel("Alcohol Content (%)")
plt.ylabel("Volatile Acidity")
plt.legend(title="Wine Quality")
plt.tight_layout()
plt.show()
```

### Categorical Relationship Analysis

The categorical analysis groups wines according to quality scores and alcohol content.

A cross-tabulation and grouped bar chart were used to compare the frequencies of different wine quality and alcohol categories.

A box plot was created to compare the distribution of alcohol content across wine quality categories. This visualization allows differences in median alcohol content, variability, and potential outliers to be examined.

Finally, a colored scatter plot was used to investigate the relationship between alcohol content and volatile acidity while distinguishing wine quality categories.

These visualizations demonstrate how categorical and numerical variables can be analyzed together to identify potential patterns.
